# CSS自定义属性

## padding 内边距

定义元素内容与边框之间的空间

单位可填写：px（像素）、em（相对单位）、%（百分比）、rem（根元素字体大小）、vw（视口宽度）、vh（视口高度）等
padding: 0px 0px 0px 0px;

## margin 外边距

定义元素与其他元素之间的空间

## front 字体相关的属性

<font color="lightcoral">font: font-style font-variant font-weight font-size/line-height font-family;</font>  
eg.font: italic small-caps bold 16px/1.5 Arial, sans-serif;

- font-style: 字体样式  
  normal（正常） italic（斜体） oblique（倾斜）
- font-variant: 字体变体  
  normal（正常） small-caps（小型大写字母）
- font-weight: 字体粗细  
  normal（正常） bold（粗体） bolder（更粗） lighter（更细） 100-900（数值）
- font-size: 字体大小  
  px（像素）、em（相对单位）、%（百分比）、rem（根元素字体大小）、vw（视口宽度）、vh（视口高度）等
- line-height:  
  行高 px（像素）、em（相对单位）、%（百分比）、rem（根元素字体大小）、vw（视口宽度）、vh（视口高度）等
- font-family: 字体族  
  serif（衬线字体） sans-serif（无衬线字体） monospace（等宽字体） cursive（手写体） fantasy（奇幻体）


## 特别的属性

- 固定在视口的特定位置，不会随页面滚动而移动  
  position: fixed;

- 显示优先级  
  z-index: 1000

- 背景
  - 背景颜色  
    background: var(--color-bg);
  - 背景模糊  
    backdrop-filter: blur(12px);
    -webkit-backdrop-filter: blur(12px);

- 边框  
  border-bottom: var(--linewidth-base-px) solid var(--color-border);


.feature-local-nav {
position: sticky;
top: calc(var(--navbar-height) + var(--title-freezed-height));
z-index: 800;
display: flex;
overflow-x: auto;
background-color: var(--color-bg);
border-left: none;
border-bottom: 1px solid var(--color-border);
}
